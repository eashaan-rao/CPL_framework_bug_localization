```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the CSV file
df = pd.read_csv('/home/cs21d002_eashaan/PhD/Objective1/results/all_project_domain_gaps.csv')
```

```python
# Sort by domain_gap_accuracy
df_sorted = df.sort_values(by='domain_gap_accuracy', ascending=True)
```

```python
df_sorted.head(20)
```

```python
# Define custom bins
bins = [0.8, 0.9, 0.91, 0.92, 0.93, 0.94, 0.95, 0.96, 0.97, 0.98, 0.99]
labels = ['0.8–0.9', '0.9–0.91', '0.91–0.92', '0.92–0.93', '0.93–0.94',
          '0.94–0.95', '0.95–0.96', '0.96–0.97', '0.97–0.98', '0.98–0.99']
```

```python
# Bin the values
df_sorted['accuracy_bin'] = pd.cut(df_sorted['domain_gap_accuracy'], bins=bins, labels=labels, right=False)
```

```python
# Count pairs per bin
bin_counts = df_sorted['accuracy_bin'].value_counts().sort_index()
```

```python
# Print counts
print("📊 Domain Gap Accuracy Distribution:")
for label, count in bin_counts.items():
    print(f"{label}: {count} pairs")
```

```python
# Plot the distribution
plt.figure(figsize=(10, 6))
bin_counts.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Distribution of Domain Gap Accuracy Ranges')
plt.xlabel('Accuracy Range')
plt.ylabel('Number of Project Pairs')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
